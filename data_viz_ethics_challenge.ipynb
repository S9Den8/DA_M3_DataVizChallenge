## Step 2: Review the BAD Charts

Below are two deliberately poor visualizations.  
Your task: Carefully review each and take notes on what you think is wrong, misleading, or just poorly designed.  
Think about chart type, colors, scales, labeling, and whether the chart tells the truth.  


```python
plt.figure(figsize=(8,5))
subset = coffee_df.groupby("Year")["Sales"].sum().reset_index()

# Wrong variable for title, truncated y-axis, misleading color & framing
plt.plot(subset["Year"], subset["Sales"], color="red", linewidth=4, marker="o")
plt.ylim(subset["Sales"].min() * 0.98, subset["Sales"].min() * 1.02)  # extreme truncation
plt.title("Shocking Collapse in Profits!", fontsize=16, fontweight="bold", color="darkred")
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
```

### Notes about bad chart #1 here:
1. The title, "Shocking Collapse in Profits!", is dramatic and charged rather than descriptive and neutral. It is biased and lacks context. I don't know what the shocking profits are referring to specifically.
2. Y-axis has been artificially truncated. This visual artificially magnifies small differences and makes them appear more dramatic than they really are. The data implies a collapse that was not supported by any data. 
3. The title describes a relationship that doesn't exist. The true variables reflect are between year and sales rather than year and profit.
4. Color is important to signify relationships. It is used to appropriately convey relationships. Red is overly dramatic and can be misunderstood as danger, failure, loss, or denote something of importance or high priority.
5. The x-axis label ("Year") and the y-axis ("Total Sales") are missing. The magnitude of sales (i.e., dollars, thousands, millions) is not indicated. This missing information distorts meaning and results in diminished understanding.
6. Scaling and notation issues exist. For example, the use of decimals along the x-axis, along with scientific notation at the top (i.e., 1e6) is unnecessary. All charts should be easy to digest and understand rather than cause cognitive strain. 
7. Chart choice is also a factor. Instead of using a line chart, which incorrectly dramatizes the relationship between variables, a bar chart would have been a better choice. A bar chart would have offered comparison between variables ("Year" x "Total Sales"). A line chart may have worked if the y-axis had started at zero, used neutral colors, the title was accurate, no heavy markers or thick lines were used. However, visually I would have preferred the bar chart.    
8. All good data sources should be cited or at least referenced so that they can be verified. Openly disclosing sources and transparency is part of ethical data practices. 
9. From a formatting perspective, I would have removed the top and right spines. I would also add padding for the title. I would consider changing or removing the red line as it is excessively thick and may amplify perceived volatility.

## Step 3: Your Turn – Make It Right

Now it’s your turn to fix things.  

- Start by thinking about a question you want to answer with this dataset.  
- Choose a visualization type that matches your question and the data structure.  
- Apply the best practices from our lessons: clear labeling, appropriate scale, purposeful color, simplicity, and accessibility.  
- Be honest - don’t manipulate the chart to make a point that the data doesn’t support.  

When you’re done, add your chart(s) below. You may create more than one if it helps tell the story better.  


```python
# GOOD visualization(s) for Chart # 1 

# Aggregate sales by year
sales_by_year = coffee_df.groupby("Year", as_index=False)["Sales"].sum()

plt.figure(figsize = (8, 5))
plt.bar(sales_by_year["Year"], sales_by_year["Sales"])

# Titles and labels 
plt.title(
    "Total Sales by Year",
    fontsize=16,
    fontweight="bold",
    pad=12
)
plt.xlabel("Year")
plt.ylabel("Total Sales (USD)")
plt.ylim(0, sales_by_year["Sales"].max() * 1.1)

# Remove spines
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Disable scientific notation on y-axis:
ax.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.figtext(
    0.01, -0.08,
    "Source: Coffee chain sales dataset",
    ha="left",
    fontsize=9
)

plt.show()

```

### Chart Improvements Explained: (Chart # 1)
- Replaced the misleading line chart with a bar chart to more accurately compare total sales by year.
- Started y-axis at zero to avoid exaggerating small differences. 
- Used a neutral title to accurately reflect the data being shown.
- Added clear axis labels and removed unnecessary scientific notation to encourage readability and avoid confusion.
- Removed decorative elements and unnecessary styling to reduce visual bias.
- Included a data source for transparency.

```python
# GOOD visualization(s) for Chart # 2 

# Aggregate average sales by region
avg_sales = coffee_df.groupby("Region", as_index=False)["Sales"].mean()

# Sort for clarity (ascending order)
avg_sales = avg_sales.sort_values("Sales")

plt.figure(figsize=(8, 5))

# Horizontal bar chart for clarity
plt.barh(
    avg_sales["Region"],
    avg_sales["Sales"],
    color="#5A8DEE"
)

# Add valye labels to bars
for i, v in enumerate(avg_sales["Sales"]):
    plt.text(
        v + 100,    # small offset to the right of the bar
        i,  # bar index
        f"${v:,.0f}",   # formatted $$$
        va="center",
        fontsize=9    
    )

# Titles and labels 
plt.title(
    "Average Coffee Sales by Region",
    fontsize=14,
    fontweight="bold",
    pad=12
)
plt.xlabel("Average Sales (USD)", fontsize=11)
plt.ylabel("Region", fontsize=11)

# Start axis as zero
plt.xlim(0, avg_sales["Sales"].max() * 1.1)

# Remove spines
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Remove gridlines 
ax.grid(False)

# Source note
plt.figtext(
    0.01, -0.08,
    "Source: Coffee chain sales dataset, regional aggregates",
    ha="left",
    fontsize=9
)

plt.tight_layout()
plt.show()
```

### Chart Improvements Explained: (Chart # 2)
-   Replaced the overly stylized vertical bar with a clean horizontal bar chart to improve comparability.
-   Removed the rainbow palette. A single neutral color was used to avoid implying meaning through color. 
-   Sorted regions in ascending order to make differences easier to interpret.
-   Added clear labels, units (USD), and geographic context.
-   Removed decorative patterns, heavy outlines, and gridlines that added no informational value. 
-   Included a data source to support transparency and ethical reporting.
-   Used average sales by region to provide a normalized comparison that is not skewed by differences in store count or volume.
